```python
df = pd.read_csv('/kaggle/input/datasets/johnsmith88/heart-disease-dataset/heart.csv')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:-1], df.iloc[:,-1], test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1 = LogisticRegression(max_iter=10000)
clf2 = DecisionTreeClassifier()
```

```python
clf1.fit(X_train,y_train)
```

```python

clf2.fit(X_train,y_train)
```

```python
y_pred1 = clf1.predict(X_test)
y_pred2 = clf2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy score of Logistic Regression: ", accuracy_score(y_test,y_pred1))
print("Accuracy score of Decision Tree: ", accuracy_score(y_test,y_pred2))
```

```python
confusion_matrix(y_test,y_pred1)
```

```python
print("Logistic Regression Confusion Matrix \n")
pd.DataFrame(confusion_matrix(y_test,y_pred1), columns=list(range(0,2)))
```

```python
print("Decision Tree Confusion Matrix \n")
pd.DataFrame(confusion_matrix(y_test,y_pred2), columns=list(range(0,2)))
```

```python
result = pd.DataFrame()
result["Actual Label"] = y_test
result['Logistic Regression Prediction'] = y_pred1
result['Decision Tree Prediction'] = y_pred2
```

```python
result.sample(8)
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score
```

```python
print("for Logistic regression Model")
print("-"*50)     #--------------------------------------------------
cdf = pd.DataFrame(confusion_matrix(y_test,y_pred1), columns=list(range(0,2)))
print(cdf)
print("-"*50)
print("Precision - ", precision_score(y_test,y_pred1))
print("Recall - ", recall_score(y_test,y_pred1))
print("F1 - ", f1_score(y_test,y_pred1))
```

```python
print("for Decision Tree Model")
print("-"*50)     #--------------------------------------------------
cdf = pd.DataFrame(confusion_matrix(y_test,y_pred2), columns=list(range(0,2)))
print(cdf)
print("-"*50)
print("Precision - ", precision_score(y_test,y_pred2))
print("Recall - ", recall_score(y_test,y_pred2))
print("F1 - ", f1_score(y_test,y_pred2))
```

```python
print(precision_score(y_test,y_pred1,average=None))
print(precision_score(y_test,y_pred2,average=None))
print(recall_score(y_test,y_pred2,average=None))
```
