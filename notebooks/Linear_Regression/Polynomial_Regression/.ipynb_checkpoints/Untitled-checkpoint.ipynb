```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression,SGDRegressor

from sklearn.preprocessing import PolynomialFeatures, StandardScaler

from sklearn.metrics import r2_score

from sklearn.pipeline import Pipeline
```

```python
X = 6 * np.random.rand(200, 1) - 3
y = 0.8 * X**2 + 0.9 * X + 2 + np.random.randn(200,1)

# y = 0.8x^2 + 0.9x + 2
```

```python
plt.plot(X, y, 'b. ') # display as the blue points
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Train Test Split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size =0.2,random_state=2)
```

```python
# Applying the linear regression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
r2_score(y_test,y_pred)
```

```python
plt.plot(X_train,lr.predict(X_train),color='r')
plt.plot(X,y,'b.')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
# Applying Polynomial Linear Regression of degree 2
poly = PolynomialFeatures(degree=2,include_bias=False)

X_train_tran = poly.fit_transform(X_train)
X_test_tran = poly.transform(X_test)
print(X_train[0])
print(X_train_tran[0])
```

```python
# including_bias parameter
```

```python
# Applying Polynomial Linear Regression of degree 2
poly = PolynomialFeatures(degree=2,include_bias=True)

X_train_trans = poly.fit_transform(X_train)
X_test_trans = poly.transform(X_test)
```

```python
print(X_train[0])
print(X_train_trans[0])
```

```python
lr = LinearRegression()
lr.fit(X_train_trans,y_train)
```

```python
y_pred = lr.predict(X_test_trans)
r2_score(y_test,y_pred)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
X_new = np.linspace(-3,3,200).reshape(200,1)
# creates 200 equally spaced numbers between -3 and 3.
X_new_poly = poly.transform(X_new)
y_new = lr.predict(X_new_poly)
```
