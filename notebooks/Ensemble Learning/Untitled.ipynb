```python
from sklearn.datasets import fetch_california_housing
import numpy as np

from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import VotingRegressor
```

```python
data = fetch_california_housing()

X = data.data
y = data.target
```

```python
# Creating 3 diff. regression model

lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
# putting them into estimator list

estimators = [
    ("lr", lr),
    ("dt", dt),
    ("svr", svr)
]
```

```python
for name, model in estimators:

    scores = cross_val_score(
        model,
        X,
        y,
        scoring="r2",
        cv=10,
        n_jobs=1
    )

    print(
        f"{name}: ",
        f"{scores.mean():.3f}"
    )
```

```python
vr = VotingRegressor(estimators= estimators)

scores = cross_val_score(vr, X, y, scoring="r2", cv =10, n_jobs=-1)
print("\nVoting Regressor")
print("-" *50)
print(f"Mean R2:{scores.mean():.3f}")
```

```python
# VR with diff weights

print("\nWeighted Voting Regressor")
print("-" * 50)

best_score = -np.inf
best_weights = None

for i in range(1, 4):

    for j in range(1, 4):

        for k in range(1, 4):

            weights = [i, j, k]

            vr = VotingRegressor(
                estimators=estimators,
                weights=weights
            )

            scores = cross_val_score(
                vr,
                X,
                y,
                scoring="r2",
                cv=10,
                n_jobs=-1
            )

            mean_score = scores.mean()

            print(
                f"Weights={weights} "
                f"R2={mean_score:.3f}"
            )

            if mean_score > best_score:
                best_score = mean_score
                best_weights = weights


print("\nBest result")
print("-" * 50)
print("Best weights:", best_weights)
print("Best CV R2:", round(best_score, 3))
```

```python
dt1 = DecisionTreeRegressor(
    max_depth=1,
    random_state=42
)

dt2 = DecisionTreeRegressor(
    max_depth=3,
    random_state=42
)

dt3 = DecisionTreeRegressor(
    max_depth=5,
    random_state=42
)

dt4 = DecisionTreeRegressor(
    max_depth=7,
    random_state=42
)

dt5 = DecisionTreeRegressor(
    max_depth=None,
    random_state=42
)
```

```python
tree_estimators = [
    ("dt1", dt1),
    ("dt2", dt2),
    ("dt3", dt3),
    ("dt4", dt4),
    ("dt5", dt5)
]
```

```python
print("\nIndividual Decision Trees")
print("-" * 50)

for name, model in tree_estimators:

    scores = cross_val_score(
        model,
        X,
        y,
        scoring="r2",
        cv=10,
        n_jobs=-1
    )

    print(
        f"{name}: "
        f"{scores.mean():.3f}"
    )
```

```python
vr_trees = VotingRegressor(
    estimators=tree_estimators
)

scores = cross_val_score(
    vr_trees,
    X,
    y,
    scoring="r2",
    cv=10,
    n_jobs=-1
)

print("\nVoting Regressor - Decision Trees")
print("-" * 50)

print(
    f"Mean R2: {scores.mean():.3f}"
)
```
