```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
import pandas as pd
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2, random_state=4)
```

```python
from sklearn.linear_model import SGDRegressor
reg = SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
```

##### SGDRegressor trains a linear regression model using stochastic gradient descent. max_iter=500 controls the number of epochs, eta0=0.1 is the learning rate, learning_rate='constant' keeps the learning rate fixed during training, penalty='l2' adds Ridge regularization(the lamda things which is added extra in the loss function), and alpha=0.001 controls the strength of that regularization to reduce overfitting.

```python
reg.fit(X_train,y_train)

y_pred = reg.predict(X_test)
print("r2_score", r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
```

```python
class MeraRidgeGD:

    def __init__(self,epochs,learning_rate,alpha):
        self.learning_rate= learning_rate
        self.epochs = epochs
        self.alpha = alpha
        self.intercept_ = None
        self.coef_ = None

    def fit(self,X_train,y_train):
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0
        thetha = np.inset(self.coef_,0 , self.intercept_)

        X_train = np.insert(X_train,0,1,axis=1)

        for i in range(self.epochs):
            thetha_der = np.dot(X_train.T,X_train).dot(thetha) - np.dot(X_train.T,y_train) + self.alpha*thetha
            # W in the maths is treated as the thetha here
            thetha = thetha - self.learning_rate*thetha_der

        self.intercept_ = thetha[0]
        self.coef_ = thetha[1:]

    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
reg = MeraRidgeGD(epochs =500,alpha=0.001,learning_rate=0.005)
```
